**1. Write a function to return the maximum of three numbers.**

```python
def max_of_nums(num1, num2, num3):
  return max(num1, num2, num3)

num_1 = int(input("Enter first number: "))
num_2 = int(input("Enter second number: "))
num_3 = int(input("Enter third number: "))

print("Maximum is ", max_of_nums(num_1, num_2, num_3))
```

**2. Create a function that returns the factorial of a number using recursion.**

```python
def factorial_num(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n * factorial_num(n-1)

num = int(input("Enter a number: "))
print(f"Factorial of {num} is:", factorial_num(num))
```

**3. Write a function that checks whether a number is prime.**

```python
def is_prime(n):
  if n <= 1:
    return False
  else:
    for i in range(2, n):
      if n % i == 0:
        return False
      return True

num = int(input("Enter a number: "))
if is_prime(num):
  print("Prime Number")
else:
  print("Not a Prime number")
```

**5. Write a function that takes a list and returns the sum of all elements.**

```python
def sum_of_list(lst):
  for i in lst:
    return sum(lst)

list_of_num = list(map(int, input("Enter a list of numbers: ").split()))
print("Sum of list is:",sum_of_list(list_of_num))
```

**6. Create a function to check if a string is a palindrome.**

```python
def is_plindrome(p):
  p = p.lower()
  return p == p[::-1]

st = input("Enter a string: ")
if is_plindrome(st):
  print("String is Palindrome")
else:
  print("String is not Plindrome")
```

**7. Write a function to calculate the area of a circle given the radius.**

```python
import math

def area_of_circle(r):
  return math.pi * (r * r)

radius = float(input("Enter radius of circle: "))
print("Area of circle is",area_of_circle(radius))
```

**8. Create a function to return the Fibonacci sequence up to `n` terms.**

```python
def fibonacci(n):
  a = 0
  b = 1
  for i in range(n):
    print(i, end=" ")
    a, b = b, a + b

num = int(input("Enter a number: "))
print("Fibonacci sequence is:")
fibonacci(num)
```

**9. Write a function that takes a list of numbers and returns only even numbers.**

```python
def even_num(n):
  for i in n:
    if i % 2 == 0:
      print(i, end=" ")

num = list(map(int, input("Enter a list of numbers: ").split()))
print("Even numbers are:")
even_num(num)
```

**10. Define a function to count the number of vowels in a given string.**

```python
def vowel(v):
  vo = "aeiouAEIOU"
  count = 0
  for char in v:
    if char in vo:
      count += 1
  return count

st = input("Enter a string: ")
print("vowels in string are:", vowel(st))
```

**12. Create a function to convert Celsius to Fahrenheit.**

```python
def far(c):
  f = ((9/5) * c) + 32
  return f

temperature = int(input("Enter temperature in fahrenheit: "))
print("Temperature in celsius is:", far(temperature))
```

**13. Define a function that accepts a sentence and returns the word with the maximum length.**

```python
def max_len(sen):
  word = sen.split()
  max_word = max(word, key = len)
  return max_word

st = input("Enter a sentence: ")
print("Word with maximum length is:", max_len(st))
```

**14. Write a function that accepts a number and returns whether it is an Armstrong number.**

```python
def is_armstrong(num):
  num_to_str = str(num)
  num_len = len(num_to_str)
  total = 0

  for digit in num_to_str:
    total += int(digit) ** num_len
  return total == num

num = int(input("Enter a number: "))
print("Number is armstrong: ",is_armstrong(num))
```
