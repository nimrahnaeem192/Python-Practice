**5. Use a `while` loop to calculate the factorial of a number entered by the user.**

```python
factorial_num = int(input("Enter a number to calculate its factorial: "))
result = 1
i = 1
while i <= factorial_num:
  result *= i
  i += 1
print(f"Fcatorial of {factorial_num} is: {result}")
```

**11. Print all prime numbers between 1 and 100 using a loop.**

```python
for num in range(2, 101):
  is_prime = True
  for i in range(2, num):
    if (num % i == 0):
      is_prime = False
      break

  if is_prime:
    print(num, end=" ")
```

**12. Ask the user for a number and use a loop to check if it is a palindrome number.**

```python
num = int(input("Enter a Number to check if its palindrome: "))
original_num = num
reverse_num = 0

while num > 0:
  digit = num % 10
  reverse_num = reverse_num * 10 + digit
  num //= 10

if (original_num == reverse_num):
  print("Its a Plindrome number")
else:
  print("Number is not a palindrome")
```

**13. Print the Fibonacci sequence up to 10 terms using a loop.**

```python
num1 = 0
num2 = 1

for i in range(10):
  print(num1, end = " ")
  next_num = num1 + num2
  num1 = num2
  num2 = next_num
```

**14. Count how many digits are in an integer entered by the user using a `while` loop.**
