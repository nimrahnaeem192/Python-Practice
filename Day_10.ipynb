**4. Create a program to count the number of words in a text file.**

```python
file = open('myfile.txt', 'r')
count = 0
for i in file:
  words = i.split()
  count  += len(words)
print("Number of words in file is:",count)
file.close()
```

**5. Write a program that reads a file and counts the number of lines in it.**

```python
file = open('myfile.txt', 'r')
count = 0
for i in file:
  count += 1
print("Number of lines in file:",count)
file.close()
```

**6. Write a program to copy the contents of one file to another.**

```python
with open('myfile.txt', 'r') as file1:
  content = file1.read()

with open('myfile2.txt', 'w') as file2:
  file2.write(content)

print("Content copied sucessfully!")
```

**7. Write a program to read a file line by line and print each line.**

```python
with open('myfile2.txt', 'r') as f:
  data = f.readlines()
  for i in data:
   print(i)
```

**8. Write a Python program to reverse the contents of a file.**

```python
with open('myfile2.txt', 'r') as file:
  content = file.read()
  print(content[::-1])
```

**9. Write a program to read a CSV file and display its content.**

```python
import csv

with open('myfile3.csv', 'w', newline= "") as f:
  write = csv.writer(f)
  write.writerow(["name", "age", "city"])
  write.writerow(["sana", "19", "lahore"])
  write.writerow(["sara", "23", "karachi"])


with open('myfile3.csv', 'r', newline= "") as file:
  data = csv.reader(file)
  for row in data:
   print(row)
```

**10. Write a program to merge the contents of two files into a third file.**

```python
with open('myfile.txt', 'r') as file1:
  content1 = file1.read()

with open('myfile2.txt', 'r') as file2:
  content2 = file2.read()

with open('myfile3.txt', 'w') as file3:
  file3.write(content1 + "\n" + content2)

print("Files merged successfully!")
```

**11. Create a program that removes all blank lines from a text file.**

```python
with open('myfile.txt', 'r') as f1:
  content = f1.readlines()

with open('newfile', 'w') as f2:
  for line in content:
    if line.strip():
      f2.write(line)
```

**13. Write a Python script to read the last n lines of a file.**

```python
with open('myfile.txt', 'r') as f:
  lines = f.readlines()
  last_line = lines[-1:]
  for line in last_line:
   print(line)
```

**14. Create a program that saves user input to a file until the user types "exit".**

```python
with open('file.txt', 'w') as f:
  while True:
    data = input("Enter data and type 'exit' to stop ")

    if data.lower() == 'exit':
     break

    f.write(data + "\n")
```

**15. Write a program to check if a file exists and print its size in bytes.**

```python
import os

with open('myfile.txt', 'r') as f:
  if os.path.exists('myfile.txt'):
    size = os.path.getsize('myfile.txt')
    print(f"Size of {'myfile.txt'} is: {size} ")
  else:
    print("File does not exist")
```
