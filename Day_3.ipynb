**12.	What is the difference between == and is in Python?**

'==' identifies if the 2 variiables have same content or not while

'is' identifies if the 2 variables have same memory location or not

```python
x = [1, 2, 3]
y = [1, 2, 3]

print(x == y) #returns True because the content of x and y is same
print(x is y ) #returns False as memory location of x and y is different

z = x
print(x is z) #returns True as memory location of x and z is same
```
