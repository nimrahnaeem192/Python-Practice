**Q10. Write a Python program that uses the True and False keywords in a** **condition.**

```python
num = int(input("Enter a num to check if it is even or not: "))
def check(num):
 if (num % 2 == 0):
  return True
 else:
  return False

result = check(num)
print(f"Number is even: {result}")
```
