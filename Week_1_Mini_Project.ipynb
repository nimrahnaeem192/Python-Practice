Personal Finance Management System (Mini-Project)



```python
print("Welcome to your Personal Finance Management System")
transactions = []
balance = 0
while True:
  print("Enter 1 to Add Income")
  print("Enter 2 to Add Expense")
  print("Enter 3 to View Transactions History")
  print("Enter 4 to View balance")
  print("Enter 5 to Exit")
  choice = int(input("Enter your choice: "))

  if (choice == 1):
    income = int(input("Enter Your Income: "))
    balance += income
    transactions.append("Income: " + str(income))

  elif (choice == 2):
    expense = int(input("Enter Your Expense: "))
    balance -= expense
    transactions.append("Expense:  " + str(expense))

  elif (choice == 3):
    if len(transactions) == 0:
      print("No Transaction History")
    else:
      print("Transaction History: ")
      for t in transactions:
        print(t)

  elif (choice == 4):
    print("Your Balance is: ", balance)

  elif (choice == 5):
    break

  else:
    print("You Enter Invalid Choice")

print("Thankyou for using Personal Finance Management System")
```
