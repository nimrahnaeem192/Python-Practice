**1.	Write a Python program to merge two dictionaries into one.**

```python
dic1 = {
    "a" : 1 ,
    "b" : 2 ,
}

dic2 = {
    "c" : 3 ,
    "d" : 4
}

dic1.update(dic2)
print(dic1)
```

**2.	How can you check if a key exists in a dictionary?**

```python
dic = {
    "a" : 1 ,
    "b" : 2
}

if "c" in dic:
  print("Exist")
else:
  print("Not Exist")
```

**5.	Convert two lists into a dictionary using zip().**

```python
keys = ["a", "b", "c"]
values = [2, 4, 6]
dic1 = dict(zip(keys, values))
print(dic1)
```

**6.	Create a dictionary from user input of 3 key-value pairs.**

```python
dic = {}
for i in range(3):
  keys = input(f"Enter key {i+1}: ")
  values = input(f"Enter Value for '{keys}': ")
  dic[keys] = values
print(dic)
```

**7.	Write a program to count the frequency of each character in a string using a dictionary.**

```python
st = "Python"
char_freq = {}
for char in st:
  if char in char_freq:
    char_freq[char] += 1
  else:
    char_freq[char] = 1
print(char_freq)
```

**8.	Given a dictionary of students and scores, find the student with the highest score.**

```python
students = {
    "Sana" : 87 ,
    "Amna" : 92 ,
    "Fatima" : 76 ,
    "Abiha" : 99
}

highest_score_student = max(students, key = students.get)
print("Top Student: ",highest_score_student)
print("Highest Score: ",students[highest_score_student])
```

**10.	Find the intersection of two sets in Python.**

```python
set1 = {1, 3, 5}
set2 = {1, 2, 3}
s = set1.intersection(set2)
print(s)
```

**11.	Write a program that takes a string and returns the number of vowels in it.**

```python
st = "Python Programming"
vowels = "aeiouAEIOU"
count = 0
for char in st:
  if char in vowels:
    count += 1
print("Vowels:", count)
```

**12.	Reverse a list without using the reverse() method.**

```python
lst = [1, 2, 3]
reverse_list = lst[::-1]
print(reverse_list)
```
